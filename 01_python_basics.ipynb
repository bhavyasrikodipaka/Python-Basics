```python
#creating a list of student names
students = ["Bhavya", "Riya", "Preeti", "Ajay"]
#printing the list
print("Initial list of students:",students)
#adding a new student
students.append("Keerthi")
print("Updated list of students:",students)
#access individual students
print("First student:",students[0])
print("Second student:",students[1])
print("Total number of students:",len(students))
```

```python
#using for loop to greet every student in the list automatically
for student in students:
  print(f"Hello {student}, welcome to Python class!")
```

```python
#a list of student marks
marks = [85, 42, 78, 33, 91]

for mark in marks:
  if mark >=50:
    print(f"Score {mark}: Passed!🎉")
  else:
    print(f"Score {mark}: Retake needed!📃")
```

```python
ages = [15, 22, 17, 30, 12]

for x in ages:
  if x >= 18:
    print("Adult, You are eligible to vote!")
  else:
    print("Minor, You are not eligible to vote!")
```
