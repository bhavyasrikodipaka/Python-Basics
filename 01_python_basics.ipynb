```python
#define a function that calculates tax and final price
def calculate_total_price(price, tax_rate=0.10):
  tax_amount = price * tax_rate
  final_price = price + tax_amount
  return final_price

#use the function multiple times with different inputs
shoe_price = calculate_total_price(100)
book_price = calculate_total_price(25)
laptop_price = calculate_total_price(1000, 0.18) #standard 18% tax

print(f"Shoes: ${shoe_price}")
print(f"Book: ${book_price}")
print(f"Laptop: ${laptop_price}")
```

```python
#a function
def evaluate_grade(name, score):
  if (score >= 90):
    return f"{name}: A Grade"
  elif (score >=75):
    return f"{name}: B Grade"
  else:
    return f"{name}: Needs Improvrmrnt"

print(evaluate_grade("Bhavya", 92))
print(evaluate_grade("Riya", 44))
print(evaluate_grade("Ajay", 10))
```

```python
#function!!
def analyze_student(name, score):
  if score >= 90:
    return f"{name}: Outstanding (Honor Roll)", True
  elif score >= 75:
    return f"{name}: Passed with Distinction", False
  elif score >= 50:
    return f"{name}: Passed", False
  else:
    return f"{name}: Retake Required", False

#dictionary
class_roster = {
    "Bhavya": 96,
    "Riya": 88,
    "Ajay": 45,
    "Keerti": 92,
    "Sneha": 78,
    "Rahul": 38
}

#list
honor_roll = []

print("--- CLASS REPORT ---")

#loop and conditionals
for student, mark in class_roster.items():
  report, is_honor = analyze_student(student, mark)
  print(report)

  #add to honor roll list if qualified
  if is_honor:
    honor_roll.append(student)

#final summary
print("--- HONOR ROLL MEMBERS ---")
print(f"Top Students: {honor_roll}")
```

```python
#with input()
def analyze_student(name, score):
    if score >= 90:
        return f"🌟 {name}: Outstanding (Honor Roll)", True
    elif score >= 75:
        return f"👍 {name}: Passed with Distinction", False
    elif score >= 50:
        return f"✅ {name}: Passed", False
    else:
        return f"📚 {name}: Retake Required", False

#dictionary
class_roster = {}
honor_roll = []

#ask how many students to enter
num_students = int(input("How many students do you want to add? "))

#loop
for i in range(num_students):
  name = input(f"Enter name for student:")
  score = int(input(f"Enter score for {name}:"))

  #store into dictionary
  class_roster[name] = score

#evaluate and display results
print("--- GENERATING CLASS REPORT ---")

for student, mark in class_roster.items():
  report, is_honor = analyze_student(student, mark)
  print(report)

  if is_honor:
    honor_roll.append(student)

print("--- HONOR ROLL MEMBERS ---")
print(f"Top Students: {honor_roll}")
```
