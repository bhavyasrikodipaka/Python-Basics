```python
#creating a list of student names
students = ["Bhavya", "Riya", "Preeti", "Ajay"]
#printing the list
print("Initial list of students:",students)
#adding a new student
students.append("Keerthi")
print("Updated list of students:",students)
#access individual students
print("First student:",students[0])
print("Second student:",students[1])
print("Total number of students:",len(students))
```

```python
#using for loop to greet every student in the list automatically
for student in students:
  print(f"Hello {student}, welcome to Python class!")
```

```python
#a list of student marks
marks = [85, 42, 78, 33, 91]

for mark in marks:
  if mark >=50:
    print(f"Score {mark}: Passed!🎉")
  else:
    print(f"Score {mark}: Retake needed!📃")
```

```python
#a dictionary of students marks
student_scores = {
    "Bhavya": 95,
    "Riya": 88,
    "Ajay": 42,
    "Keerti": 76
}
#accessing a score using the student's name (key)
print("Bhavya's score is:",student_scores["Bhavya"])
#new student
student_scores["Sneha"] = 91
#printing all key-value pairs using a loop
for x, y in student_scores.items():
  print(f"{x} scored {y}%")
```

```python
#check who passed and who failed directly from the dictionary
for x, y in student_scores.items():
  if y >= 50:
    print(f"{x} passed with {y}%")
  else:
    print(f"{x} scored {y}% and needs to retake.")
```

```python
#define a function that calculates tax and final price
def calculate_total_price(price, tax_rate=0.10):
  tax_amount = price * tax_rate
  final_price = price + tax_amount
  return final_price

#use the function multiple times with different inputs
shoe_price = calculate_total_price(100)
book_price = calculate_total_price(25)
laptop_price = calculate_total_price(1000, 0.18) #standard 18% tax

print(f"Shoes: ${shoe_price}")
print(f"Book: ${book_price}")
print(f"Laptop: ${laptop_price}")
```

```python
#a function
def evaluate_grade(name, score):
  if (score >= 90):
    return f"{name}: A Grade"
  elif (score >=75):
    return f"{name}: B Grade"
  else:
    return f"{name}: Needs Improvrmrnt"

print(evaluate_grade("Bhavya", 92))
print(evaluate_grade("Riya", 44))
print(evaluate_grade("Ajay", 10))
```
